```python
# Step 2B.1 — Create section-based chunks

def create_chunks(text):
    chunks = []
    
    current_chunk = ""
    
    for line in text.split("\n"):
        line = line.strip()
        
        if not line:
            continue
        
        # Start a new chunk when a numbered section begins
        if line[:2].replace(".", "").isdigit() or line[:3].replace(".", "").isdigit():
            
            if current_chunk:
                chunks.append(current_chunk.strip())
            
            current_chunk = line
        
        else:
            current_chunk += " " + line
    
    if current_chunk:
        chunks.append(current_chunk.strip())
    
    return chunks
```

```python
# PHASE 7C — Build the Agent Decision Logic

def contract_review_agent(review_question):

    # Step 1: Create query embedding
    query_embedding = embedding_model.encode(
        [review_question],
        convert_to_numpy=True
    ).astype("float32")

    # Step 2: Retrieve relevant policy evidence
    policy_distances, policy_indices = policy_index.search(
        query_embedding,
        k=3
    )

    retrieved_policy_context = "\n\n".join(
        policy_section_chunks[index]
        for index in policy_indices[0]
    )

    # Step 3: Retrieve relevant contract evidence
    contract_distances, contract_indices = contract_index.search(
        query_embedding,
        k=3
    )

    retrieved_contract_context = "\n\n".join(
        contract_section_chunks[index]
        for index in contract_indices[0]
    )

    # Step 4: Ask Gemini to analyze the evidence
    agent_prompt = f"""
You are an AI Contract Review Agent.

Your task is to review the contract against the provided company policy.

Use ONLY the evidence provided below.
Do not assume facts that are not present in the evidence.

REVIEW QUESTION:
{review_question}

CONTRACT EVIDENCE:
{retrieved_contract_context}

POLICY EVIDENCE:
{retrieved_policy_context}

Analyze the evidence and provide:

1. Contract Payment Term
2. Policy Standard
3. Compliance Finding
4. Reason
5. Decision Status
6. Recommended Next Step

For Decision Status, use exactly one of:
- DEVIATION_FOUND
- NO_DEVIATION
- INSUFFICIENT_EVIDENCE

For Recommended Next Step, use:
- HUMAN_REVIEW_REQUIRED
- STANDARD_REVIEW_PROCESS
- ADDITIONAL_INFORMATION_REQUIRED

Final legal or business decisions must remain with an authorized human reviewer.
"""

    # Step 5: Generate agent response
    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=agent_prompt
    )

    # Step 6: Return the agent result
    return {
        "review_question": review_question,
        "contract_evidence": retrieved_contract_context,
        "policy_evidence": retrieved_policy_context,
        "agent_review": response.text
    }
```

```python
agent_result = contract_review_agent(
    "Does this contract comply with the standard payment policy?"
)

print("===== AGENT REVIEW =====")
print(agent_result["agent_review"])
```

```python
# 7D — Make the Agent Output Structured & Workflow-Ready

# PHASE 7D — Structured Agent Output

def parse_agent_decision(agent_text):

    text = agent_text.upper()

    if "DEVIATION_FOUND" in text:
        decision_status = "DEVIATION_FOUND"
    elif "NO_DEVIATION" in text:
        decision_status = "NO_DEVIATION"
    elif "INSUFFICIENT_EVIDENCE" in text:
        decision_status = "INSUFFICIENT_EVIDENCE"
    else:
        decision_status = "UNKNOWN"

    if "HUMAN_REVIEW_REQUIRED" in text:
        next_step = "HUMAN_REVIEW_REQUIRED"
    elif "STANDARD_REVIEW_PROCESS" in text:
        next_step = "STANDARD_REVIEW_PROCESS"
    elif "ADDITIONAL_INFORMATION_REQUIRED" in text:
        next_step = "ADDITIONAL_INFORMATION_REQUIRED"
    else:
        next_step = "UNKNOWN"

    return {
        "decision_status": decision_status,
        "next_step": next_step,
        "agent_review": agent_text
    }
    
```

```python
structured_result = parse_agent_decision(
    agent_result["agent_review"]
)

print("===== STRUCTURED AGENT OUTPUT =====")
print("Decision Status:", structured_result["decision_status"])
print("Next Step:", structured_result["next_step"])
```

```python
# 7E — Connect the Agent Steps into One End-to-End Workflow.
# PHASE 7E — End-to-End Contract Review Agent

def run_contract_review_agent(review_question):

    print("===== CONTRACT REVIEW AGENT =====")
    print("Review Question:", review_question)

    # STEP 1 — Run the agent
    agent_result = contract_review_agent(review_question)

    # STEP 2 — Structure the decision
    structured_result = parse_agent_decision(
        agent_result["agent_review"]
    )

    # STEP 3 — Display decision
    print("\n===== AGENT DECISION =====")
    print("Decision Status:", structured_result["decision_status"])
    print("Next Step:", structured_result["next_step"])

    # STEP 4 — Return complete result
    return {
        "review_question": review_question,
        "contract_evidence": agent_result["contract_evidence"],
        "policy_evidence": agent_result["policy_evidence"],
        "agent_review": structured_result["agent_review"],
        "decision_status": structured_result["decision_status"],
        "next_step": structured_result["next_step"]
    }
```

```python
# Step 1 — Define the reviewer decisions
#PHASE 8A — HUMAN-IN-THE-LOOP

Human Review Process

1. Agent identifies a potential deviation.
2. Agent provides supporting contract and policy evidence.
3. Authorized reviewer reviews the finding.
4. Reviewer selects one of the following decisions:

   - APPROVE_EXCEPTION
   - REQUEST_FURTHER_REVIEW
   - REJECT_EXCEPTION

5. Reviewer decision is recorded for audit purposes.

Important:
The AI agent recommends and supports the review.
The final legal/business decision remains with the authorized human reviewer.
```

```python
# STEP 2 HUMAN REVIEW DECISIONS

The authorized reviewer can select:

• APPROVE_EXCEPTION — Approve the identified exception.
• REQUEST_FURTHER_REVIEW — Request additional information or specialist review.
• REJECT_EXCEPTION — Reject the proposed exception.

The AI agent provides recommendations and evidence, while the final
legal/business decision remains with the authorized human reviewer.
```

```python
#Step 3 — Create a simple reviewer function
# PHASE 8A — Human Review Function

def human_review(agent_result, reviewer_decision, reviewer_name="Authorized Reviewer"):

    valid_decisions = [
        "APPROVE_EXCEPTION",
        "REQUEST_FURTHER_REVIEW",
        "REJECT_EXCEPTION"
    ]

    if reviewer_decision not in valid_decisions:
        raise ValueError(
            "Invalid reviewer decision. "
            "Choose APPROVE_EXCEPTION, REQUEST_FURTHER_REVIEW, "
            "or REJECT_EXCEPTION."
        )

    review_record = {
        "reviewer": reviewer_name,
        "agent_decision_status": agent_result["decision_status"],
        "agent_recommended_next_step": agent_result["next_step"],
        "reviewer_decision": reviewer_decision
    }

    return review_record
```

```python
# step 4 : test it 
human_decision = human_review(
    final_agent_result,
    "REQUEST_FURTHER_REVIEW"
)

print("===== HUMAN REVIEW RESULT =====")

for key, value in human_decision.items():
    print(f"{key}: {value}")
```

```python
# 8B — Connect the Human Decision Back to the Workflow
# Step 1 — Create the workflow response
# PHASE 8B — Connect Human Decision to Workflow

def process_human_decision(review_record):

    decision = review_record["reviewer_decision"]

    if decision == "APPROVE_EXCEPTION":
        workflow_action = "EXCEPTION_APPROVED"
        status = "APPROVED"

    elif decision == "REQUEST_FURTHER_REVIEW":
        workflow_action = "ADDITIONAL_REVIEW_REQUIRED"
        status = "PENDING_REVIEW"

    elif decision == "REJECT_EXCEPTION":
        workflow_action = "EXCEPTION_REJECTED"
        status = "ACTION_REQUIRED"

    else:
        workflow_action = "INVALID_DECISION"
        status = "ERROR"

    return {
        "reviewer": review_record["reviewer"],
        "reviewer_decision": decision,
        "workflow_action": workflow_action,
        "status": status
    }
```

```python
# 10F.3 — Step 2: Create the Standalone Python Runner
 we'll replace the temporary runner with a version that can actually run outside Jupyter.
```

```python
runner_code = '''
import sys
import json
import os
import pickle
import faiss
from sentence_transformers import SentenceTransformer
from google import genai

# -----------------------------
# 1. Read review question
# -----------------------------
review_question = sys.argv[1]

# -----------------------------
# 2. Load saved RAG components
# -----------------------------
with open("contract_section_chunks.pkl", "rb") as f:
    contract_section_chunks = pickle.load(f)

with open("policy_section_chunks.pkl", "rb") as f:
    policy_section_chunks = pickle.load(f)

contract_index = faiss.read_index("contract_index.faiss")
policy_index = faiss.read_index("policy_index.faiss")

# -----------------------------
# 3. Load embedding model
# -----------------------------
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# -----------------------------
# 4. Connect to Gemini
# -----------------------------
client = genai.Client(
    api_key=os.environ["GEMINI_API_KEY"]
)

# -----------------------------
# 5. Retrieve policy evidence
# -----------------------------
query_embedding = embedding_model.encode(
    [review_question],
    convert_to_numpy=True
).astype("float32")

policy_distances, policy_indices = policy_index.search(
    query_embedding,
    k=3
)

retrieved_policy_context = "\\n\\n".join(
    policy_section_chunks[index]
    for index in policy_indices[0]
)

# -----------------------------
# 6. Retrieve contract evidence
# -----------------------------
contract_distances, contract_indices = contract_index.search(
    query_embedding,
    k=3
)

retrieved_contract_context = "\\n\\n".join(
    contract_section_chunks[index]
    for index in contract_indices[0]
)

# -----------------------------
# 7. Build agent prompt
# -----------------------------
agent_prompt = f"""
You are an AI Contract Review Agent.

Your task is to review the contract against the provided company policy.

Use ONLY the evidence provided below.
Do not assume facts that are not present in the evidence.

REVIEW QUESTION:
{review_question}

CONTRACT EVIDENCE:
{retrieved_contract_context}

POLICY EVIDENCE:
{retrieved_policy_context}

Analyze the evidence and provide:

1. Contract Payment Term
2. Policy Standard
3. Compliance Finding
4. Reason
5. Decision Status
6. Recommended Next Step

For Decision Status, use exactly one of:
- DEVIATION_FOUND
- NO_DEVIATION
- INSUFFICIENT_EVIDENCE

For Recommended Next Step, use:
- HUMAN_REVIEW_REQUIRED
- STANDARD_REVIEW_PROCESS
- ADDITIONAL_INFORMATION_REQUIRED

Final legal or business decisions must remain with an authorized human reviewer.
"""

# -----------------------------
# 8. Generate Gemini response
# -----------------------------
response = client.models.generate_content(
    model="gemini-3.8-flash",
    contents=agent_prompt
)

agent_text = response.text

# -----------------------------
# 9. Parse decision
# -----------------------------
text = agent_text.upper()

if "DEVIATION_FOUND" in text:
    decision_status = "DEVIATION_FOUND"
elif "NO_DEVIATION" in text:
    decision_status = "NO_DEVIATION"
elif "INSUFFICIENT_EVIDENCE" in text:
    decision_status = "INSUFFICIENT_EVIDENCE"
else:
    decision_status = "UNKNOWN"

if "HUMAN_REVIEW_REQUIRED" in text:
    next_step = "HUMAN_REVIEW_REQUIRED"
elif "STANDARD_REVIEW_PROCESS" in text:
    next_step = "STANDARD_REVIEW_PROCESS"
elif "ADDITIONAL_INFORMATION_REQUIRED" in text:
    next_step = "ADDITIONAL_INFORMATION_REQUIRED"
else:
    next_step = "UNKNOWN"

# -----------------------------
# 10. Return JSON for n8n
# -----------------------------
output = {
    "review_question": review_question,
    "decision_status": decision_status,
    "next_step": next_step,
    "agent_review": agent_text,
    "contract_evidence": retrieved_contract_context,
    "policy_evidence": retrieved_policy_context
}

print(json.dumps(output))
'''

with open("contract_review_runner.py", "w", encoding="utf-8") as file:
    file.write(runner_code)

print("Standalone contract_review_runner.py created successfully!")
```
